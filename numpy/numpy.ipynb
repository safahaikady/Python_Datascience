# Numpy array and bascis

```python
import numpy as np 
```

Array Creation

```python
#1d array creation
arr_1d=np.array([10,20,30,40])
print(arr_1d,type(arr_1d))

#2d array creation
arr_2d=np.array([[10,20,30],
                 [40,50,60],
                 [70,50,2]
                 ])
print(arr_2d)
```

List vs NumPy Array

```python
#Ap1
#List
cost_list=[10,20,30,40]
print("Selling list",cost_list*2)

#numpy array
cost_array=np.array([10,20,30,40])
print("Selling list",cost_array*2)

#Ap2
list_d1=[10,20,50,60]
list_d2=[10,20,30,25]
print("total_d",list_d1+list_d2)

array_d1=np.array([10,20,50,60])
array_d2=np.array([10,20,30,25])
print("total_d",array_d1+array_d2)
```

using an array is faster then list

```python
import time as t
#list
start=t.time() #time() is to assign current time to the variable
list_time=[i*2 for i in range(1000000)]
end=t.time()
print("list operation:",end-start)

#array
start=t.time()
array_time=np.arange(10)*2 #arange() will the numbers with in range like arange(5)=[0,1,2,3,4]
end=t.time()
print("Array operation:",end-start)
```

creating array from scratch

```python
#zeros
zeros=np.zeros((2,3))
print("Zeros matrix: \n",zeros)

#ones
ones=np.ones((3,2))
print("Ones matrix: \n",ones)

#Full
full=np.full((3,3),7)# 3*3 matrix with all element are 7
print("Full with num matrix:\n",full)

#Random 
random=np.random.random((2,2)) #first random is class net random is function to generate random number
print("Random matrix (0 to 1):\n",random)

#Sequence start,stop,step
seq=np.arange(10,31,2)
print("Sequence within range:\n",seq)
```

Vector,Matrix and tensor

```python
vector=np.array([10,20,30,40,50])
print("Vectoe:\n",vector)

matrix=np.array([[1,2,3],
                 [2,3,5]])
print("Matrix:\n",matrix)

tensor=np.array([
    [[1,2,3],[5,8,9]],
    [[2,8,7],[9,8,5]]
])
print("Tensor:\n",tensor)
```

Array Properties

```python
arr=np.array([[1,2,3],
              [4,5,6]])
print("Shape: ",arr.shape)
print("Dimension:",arr.ndim)
print("Size:",arr.size)
print("data type:",arr.dtype)
```

Array reshaping

```python
arr=np.arange(12)
print("array:",arr)
reshape=arr.reshape((3,4)) # it wil change the dimension
print("after reshape:\n",reshape)

fl=reshape.flatten()
print("Flatten operation:",fl)
```
