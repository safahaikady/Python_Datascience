### dealing with duplicate values and missing

```python
import pandas as pd
import numpy as np
```

duplicate values


```python
#overall duplicate
cdata=pd.read_csv("subdata_pandas/company.csv")
cdata
cdata.duplicated()
cdata[cdata.duplicated()]

#Based on
cdata['EEID'].duplicated() #boolean
cdata[cdata['EEID'].duplicated()] #rows with duplicate EEID

cdata.drop_duplicates() #drop overall duplicate
cdata['EEID'].drop_duplicates() #drop duplicate based on EEID
```

Missing data
