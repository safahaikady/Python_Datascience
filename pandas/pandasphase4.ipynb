## Groupby and MergeOrJoin

```python
import pandas as pd
```

### Group By

```python
exp=pd.read_excel("subdata_pandas/expense.xlsx")
exp
```

```python
#find the total espense for each category
exp.groupby('Category')['Amount'].sum() #m1
exp.groupby('Category')['Amount'].agg(['sum'] )#m1

#Give a complete expense summary for each category
exp.groupby('Category')['Amount'].agg(
    ['sum','count','mean','min','max']
)

#How much money paid by each payment mode
exp.groupby('Payment Mode')['Amount'].agg(['sum'])

#How many transaction were made using each payment mode
exp.groupby('Payment Mode')['Amount'].agg(['count'])

#For each payment mode,calculate the total amount spent across categories
exp.groupby(['Payment Mode','Category'])['Amount'].agg(['sum','count'])
exp.groupby(['Category','Payment Mode'])['Amount'].agg(['sum','count'])

#Which category have the highest total expenses
exp.groupby('Category')['Amount'].sum().sort_values(ascending=False)
```

```python
#---------------Task-----------------------
# Create a DataFrame containing the total expense of each category.
total_expense = exp.groupby('Category')['Amount']. agg(['sum'])
total_expense
# What is the average transaction amount for each payment mode?
exp.groupby('Payment Mode')['Amount']. agg(['mean'])
# Find the highest individual transaction in each category
exp.groupby('Category')['Amount']. agg(['max'])
# # Show total, average, minimum and maximum transaction amount for each payment mode.
exp.groupby('Payment Mode')['Amount'].agg(
    ['sum', 'mean','min','max']
)
```

### MergeOrJoin

```python
customer=pd.read_csv("subdata_pandas/customer.csv")
customer
payment=pd.read_csv("subdata_pandas/payment.csv")
payment
```

```python
#Innerjoin
pd.merge(left=customer, right=payment, on='customerid',how='inner')

#left join
pd.merge(left=customer, right=payment, on='customerid',how='left')

#right join
pd.merge(left=customer, right=payment, on='customerid',how='right')

#cross join
customer.merge(payment, how='cross')
```

### Compare

```python
shop_7sep={
    "Fruits":['Mango','Apple','Banana','Papaya'],
    "Price":[100,150,60,30],
    "Qunt":[10,8,12,14]
}
stock_7sep=pd.DataFrame(shop_7sep)
shop_8sep={
    "Fruits":['Mango','Apple','Banana','Papaya'],
    "Price":[120,200,60,35],
    "Qunt":[15,6,24,17]
}
stock_8sep=pd.DataFrame(shop_8sep)

#-----------------------Compare-------------------------
stock_7sep.compare(stock_8sep)
stock_7sep.compare(stock_8sep,align_axis=0)
stock_7sep.compare(stock_8sep,keep_shape=True)
stock_7sep.compare(stock_8sep,keep_equal=True)
stock_7sep.compare(stock_8sep,align_axis=0,keep_shape=True,keep_equal=True)
```

### Pivoting

```python
monthly_sales={
    "Product":['Laptop','Mobile','Laptop','Mobile','Laptop','Laptop'],
    "Month":["Jan","Jan","Feb","Feb","Mar","Mar"],
    "Price":[95000,25000,67000,12000,65000,75000]
}
sales=pd.DataFrame(monthly_sales)
#--------------------------Pivot-------------------
sales.pivot(
columns="Month",index="Product",values="Price"
)
#--------------------------Pivot table--------------
sales.pivot_table(
    columns="Month",
    index="Product",
    values="Price",
    aggfunc="sum"
)
```

### Melting

```python
shop_sales={
    "Product":['Laptop','Mobile','Mouse'],
    "Jan":[50000,25000,2000],
    'Feb':[45000,20000,1500],
    "March":[55000,32000,2500]
}
sales_df=pd.DataFrame(shop_sales)
sales_df.melt(
    id_vars="Product",
    var_name="Month",
    value_name="Price"
)
```
