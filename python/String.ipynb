String indexing and slicing

```python
#indexing........
str="DARKCHOCOLATE"
print("org:",str)
#idexing types
#Idx +ve
print("5th:",str[5])#retrives the 5th index element in the str variable
print("last element:",str[len(str)-1])#retrives the last index element in the str variable .... len() used to find length of string
#idx -ve
print("last element:",str[-1]) #retrives last element
print("first element:",str[-len(str)]) #retrives first element

#slicing........
#syntax: String_name[start:stop:step]
#default values: start=0,stop=len(string_name),step=1
print(str[0:4]) #step=1 , output:DARK
print(str[:4])#step=1,start=0 output:DARK
print(str[4:len(str)])#step=1  output:CHOCOLATE
print(str[4:]) #step=1,stop=len(str) output:CHOCOLATE
print(str[:]) #output:DARKCHOCOLATE
#reverse 
print(str[12:3:-1])
print(str[::-1]) #reverse string
print(str[4::-1])
```

String function

```python
s="raju told me that he is coder"
print(s+" ,Is it real?") #concetanation
print(len(s))#length of string
print(s.endswith("er")) #is Strign==ng end with the substring
print(s.replace("he is","I am")) #replace old with new
print(s.find("a")) #first index of substring
print(s.count("e")) #freq of the substring
print(s.capitalize()) #Capitalize the first letter of the string
print(s.upper()) #to upper case
print(s.lower()) #to lower case
```
