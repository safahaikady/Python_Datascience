list functions.....

```python
colors=["red","yellow"]
print("org list:",colors)

print(len(colors)) #length of list

colors.append("gold") #insert the new element in the end
print("After adding gold:",colors)

colors[1]="purple"#replace an element
print("after replace")
print(colors)

colors.insert(2,"pink") #Insert the value at particular index
print("After inserting: ",colors)

colors.remove("pink")#remove an element
print("after removing element: ",colors)

colors.pop() #remove last element
print("After pop function:",colors)

```

-----------------------Tuple------------------------

```python
#it is the collection of element in the python that is ordered(insertion).
#immutable
#duplicate values are allowed
num=(10,20,23,20)
print(num) 

#Types:
#type1: using parenthesis
a=(1,2,3,4)
b=("red","blue")
mixed=("safa",19,23.007,True)
nested=(1,2,("raj",5),[12,3,6])

#type2:
#without parenthesis are also allowed
num1=10,20,23
print(num1)

#type3: using tuple constructor
val=tuple((1,2,300))
list_item=[1,2,8,9]
tuple_item=tuple(list_item)

#type4
#single element tuple
n=(10,)# for single element we have to give the coma after the element
print(n) 

print(len(n))
```

tuple functions...............

```python
#slicing
numbers = (10, 20,20, 30, 40, 50)
print(numbers[1:4]) #start=1,stop=4
print(numbers[:3]) #start=0,stop=3
print(numbers[2:])#start=2,stop=length of tuple

#count():frequency of the element
print(numbers.count(20))

#index():index of first occurrence
print(numbers.index(20))

#max():Returns the largest value
print(max(numbers))

#min():Returns the smallest value.
print(min(numbers))

#sum():Adds the elements.
print(sum(numbers))

#sorted(): Returns a list, not a tuple.
print(sorted(numbers))

#Because tuples cannot be modified after creation(immutable).If you need to change the data, you generally create a new tuple
numbers1 = (10, 20, 30)
numbers1 = (100, 20, 30)
print(numbers1)

#Or convert tuple to list for modification
list_item=list(numbers)
list_item[2]=15
tuple_item=tuple(list_item)
print(tuple_item)
```

-----------------------SET------------------------------


```python
#set  is written in {}
#set can't have duplicate values
fruits={'apple','banana','kiwi','kiwi'}
print(fruits)

#set doesn't support indexing . we can use loop for accessing elements
for fruit in fruits:
    print(fruit,end=" ")
print()

#set doesn't support slicing
#we can convert set to list for slicing
numbers1=list(fruits)
print(numbers1[::-1])


```

set functions.....


```python
#add a element
fruits={'apple','banana','kiwi','kiwi'}
fruits.add('orange')
print(fruits)

#add multiple elements
fruits.update(['guava','grapes'])
print(fruits)

#remove element
fruits.remove('apple')
print(fruits) #gives an error if element doeb\n't exist

#discard a element
fruits.discard('apple')
print(fruits) #doesn't gives an error if element doesn't exist

#removes all elements
fruits.clear()
print(fruits)

print(len(fruits))

```

set operations........

```python
#combines two sets
fruits={'apple','banana','kiwi','kiwi'}
fruits1={'guava','grapes','kiwi'}
print(fruits | fruits1)

#intersection
print(fruits & fruits1)

#difference:returns elements present in fruits but not in fruits1
print(fruits-fruits1)
```

dictionary functions and methods...............

```python
student={
    'id':"24cs118",
    'name':"safa",
    'age':20,
    'dept':"CSE"
}
#retrive single value using key
get_value=student.get('age')
print(get_value)

#check all the keys
dict_keys=student.keys()
print(dict_keys)

#retrive all the values
dict_value=student.values()
print(dict_value)

#update
update_age=student.update({"age":19})
print(student)

#set default
dep_val=student.setdefault("dept","cs")
print(dep_val)
new_loc=student.setdefault("place","udupi")
print(student)

#copy
c=student.copy()
print(c)

#pop-remove key-value based on key
student.pop("dept")
print(student)

#popitem-remove last entered element
student.popitem()
print(student)

#list form
print(student.items())
```

dict iterate................................

```python
student={
    'id':"24cs118",
    'name':"safa",
    'age':20,
    'dept':"CSE"
}
#iterate over keys
for k in student:
    print(k)

#iterate over values
for v in student.values():
    print(v)

#iterate over key-values
for k,v in student.items():
    print(k,":",v)
```
