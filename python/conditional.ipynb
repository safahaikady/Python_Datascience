Python if-else

```python
age=int(input("Enter thhe age: "))
if age>=18:
    print("Allowed for toxic")
else:
    print("not allowed for toxic")
```

if-elif-else

```python
age=int(input("Enter the age: "))
if age<=3:
    print("Child")
elif age>=4 and age<=12:
    print("Kid")
elif age>=13 and age<=18:
    print("teenage")
elif age>=19 and age<=60:
    print("Adult")
else:
    print("Old")
```

Short Hand


```python
#one line if else
a=2
b=3
print("A") if a>b else print("B")

#one line assigning
a=3
b=13
greater=a if a>b else b
print("greater is:",greater)

#even or odd example
num=25
print("Even") if num%2==0 else print("odd")
```
