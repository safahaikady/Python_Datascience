for loop

```python
#sytax: for var_name in range(start,stop,step)
#example 1
n=10
for i in range(n): #stop=n which is excluded , start=0 , step=1
    print(i,end=" " ) #end is for one line output

print()

#example 2
for i in range(1,10,2): #start=1,stop=10,step=2
    print(i, end=" ") 

```

While-loop


```python
#syntax: while condition: #statement
n=3876
count=0
while n>0:
    n//=5
    count+=1
print(count)
```

```python
n=int(input("Enter n:"))
m=int(input("Enter m:"))
for i in range(1,m+1):
    print(i,"*",n,"=",i*n) #print(f"{i}*{n}={i*n}")
```
