### MAD

```python
IT_Department_A = {
    "Employee ID": ["E101", "E102", "E103", "E104", "E105", "E106"],
    "Employee Name": ["Rohan", "Kiran", "Aditya", "Sneha", "Pooja", "Rahul"],
    "Project": [75, 65, 73, 68, 72, 67]
}

IT_Department_B= {
    "Employee ID": ["E201", "E202", "E203", "E204", "E205", "E206"],
    "Employee Name": ["Arjun", "Deepak", "Kavya", "Megha", "Sanjay", "Varun"],
    "Project": [90, 47, 93, 47, 96, 51]
}
deptA_df = pd.DataFrame(IT_Department_A)
deptB_df = pd.DataFrame(IT_Department_B)
```

```python
plt.scatter([i for i in range(0,len(deptA_df['Project']))],deptA_df['Project'],s=100,c='red',label='Dept A')
plt.scatter([i for i in range(0,len(deptB_df['Project']))],deptB_df['Project'],s=100,c='green',label='Dept B')
meanVal=np.mean(deptA_df['Project'])
plt.axhline(y=meanVal,color='blue',linestyle='--',label='Mean')
plt.ylim(30,110)
plt.xlabel("Employee ID")
plt.ylabel("Project")
plt.legend()
plt.show()
```

```python
#Find MAD IT_Department_A
x_i=deptA_df['Project']
x_bar_mean=np.mean(deptA_df['Project'])
n=len(deptA_df['Project'])
deptA_MAD=np.sum(np.abs(x_i-x_bar_mean))/n
print("DeptA MAD: ",deptA_MAD)

#Find MAD IT_Department_B
x_i=deptB_df['Project']
x_bar_mean=np.mean(deptB_df['Project'])
n=len(deptB_df['Project'])
deptB_MAD=np.sum(np.abs(x_i-x_bar_mean))/n
print("DeptB MAD: ",deptB_MAD)
```

```python
df=pd.read_csv('student_scores.csv')
df=df.head(100)
groupA = df[df['EthnicGroup'] == 'group A']
groupB = df[df['EthnicGroup'] == 'group B']
plt.scatter([i for i in range(0,len(groupA['MathScore']))],groupA['MathScore'],s=100,c='red',label='Group A')
plt.scatter([i for i in range(0,len(groupB['MathScore']))],groupB['MathScore'],s=100,c='green',label='Group B')
meanVal=np.mean(df['MathScore'])
plt.axhline(y=meanVal,color='blue',linestyle='--',label='Mean')
plt.ylim(30,110)
plt.xlabel("Employee ID")
plt.ylabel("Math Score")
plt.legend()
plt.show()
```

```python
#Find MAD GroupA
x_i=groupA['MathScore']
x_bar_mean=np.mean(groupA['MathScore'])
n=len(groupA['MathScore'])
groupA_MAD=np.sum(np.abs(x_i-x_bar_mean))/n
print("GroupA MAD: ",groupA_MAD)

#Find MAD GroupB
x_i=groupB['MathScore']
x_bar_mean=np.mean(groupB['MathScore'])
n=len(groupB['MathScore'])
groupB_MAD=np.sum(np.abs(x_i-x_bar_mean))/n
print("GroupB MAD: ",groupB_MAD)
```
