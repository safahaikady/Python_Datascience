### Percentiles

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic_df=pd.read_csv('titanic.csv')
titanic_df.head(3)
```

```python
#Data
```

```python
#replace NaN values with mean of Age
mean_age=titanic_df['Age'].mean()
titanic_df['Age']=titanic_df['Age'].fillna(mean_age)
```

```python
#for age find percentiles
print("0%(min):",np.percentile(titanic_df['Age'],0))
print("25%(Q1):",np.percentile(titanic_df['Age'],25))
print("50%(median):",np.percentile(titanic_df['Age'],50))
print("75%(Q3):",np.percentile(titanic_df['Age'],75))
print("100%(max):",np.percentile(titanic_df['Age'],100))
```

```python
sns.boxplot(x='Age',data=titanic_df)
plt.show()
```

```python
#replace NaN values with mean of Fare
mean_fare=titanic_df['Fare'].mean()
titanic_df['Fare']=titanic_df['Fare'].fillna(mean_fare)
```

```python
# For Fare find percentiles
print("0%(min):",np.percentile(titanic_df['Fare'],0))
print("25%(Q1):",np.percentile(titanic_df['Fare'],25))
print("50%(median):",np.percentile(titanic_df['Fare'],50))
print("75%(Q3):",np.percentile(titanic_df['Fare'],75))
print("100%(max):",np.percentile(titanic_df['Fare'],100))
```

```python
sns.boxplot(x='Fare',data=titanic_df)
plt.show()
```
