### <i>Measure of Central Tendency</i>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#load dataset
titanicdf = pd.read_csv('titanic.csv')
titanicdf.head(10)
```

### <u>Mean</u>

```python
#titanicdf['Fare'].mean()
meanVal=np.mean(titanicdf['Fare'])
meanVal
```

```python
sns.histplot(x=titanicdf['Fare'], data=titanicdf,bins=[i for i in range(0, 151, 10)])
plt.plot([meanVal for i in range(0,351)], [i for i in range(0,351)], c='red')
plt.ylim(0, 400)
plt.show()
```
