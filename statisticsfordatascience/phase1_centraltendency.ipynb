### <i>Measure of Central Tendency</i>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
%pip install seaborn
```

```python
#load dataset
titanicdf = pd.read_csv('titanic.csv')
titanicdf.head(10)
```

### <u>Mean</u>

```python
#titanicdf['Fare'].mean()
meanVal=np.mean(titanicdf['Fare'])
meanVal
```

```python
sns.histplot(x=titanicdf['Fare'], data=titanicdf,bins=[i for i in range(0, 151, 10)])
plt.plot([meanVal for i in range(0,351)], [i for i in range(0,351)], c='red')
plt.ylim(0, 400)
plt.show()
```

### <u>Median</u>

```python
medianVal=np.median(titanicdf['Fare'])
medianVal
```

```python
sns.histplot(x=titanicdf['Fare'], data=titanicdf,bins=[i for i in range(0, 151, 10)],color='red')
plt.plot([medianVal for i in range(0,351)], [i for i in range(0,351)], c='blue')
plt.ylim(0, 400)
plt.show()
```

### <u>Mode</u>

```python
import statistics as st
modeVal=st.mode(titanicdf['Fare'])
modeVal
```

```python
sns.histplot(x=titanicdf['Fare'], data=titanicdf,bins=[i for i in range(0, 151, 10)],color='orange')
plt.plot([modeVal for i in range(0,351)], [i for i in range(0,351)], c='green')
plt.ylim(0, 400)
plt.show()
```

### <u>Final Output</u>

```python
sns.histplot(x=titanicdf['Fare'], data=titanicdf,bins=[i for i in range(0, 151, 10)],color='pink')
plt.plot([meanVal for i in range(0,351)], [i for i in range(0,351)], c='red',label='Mean')
plt.plot([medianVal for i in range(0,351)], [i for i in range(0,351)], c='blue',label='Median')
plt.plot([modeVal for i in range(0,351)], [i for i in range(0,351)], c='green',label='Mode')
plt.ylim(0, 400)
plt.legend()
plt.title('Titanic Ticket Fare Central Tendency',fontsize=15,fontweight='bold')
plt.tight_layout()
plt.show()
```
