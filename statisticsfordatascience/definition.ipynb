## Population:<br>
* A population includes of all the elements from a set of data.the population is the whole set of values or individuals,you are intrested in.<br>
* Population characteristics are mean,standard deviation,proportion,median,percentiles etc.<br>


#### Sample(N):<br>
* A sample is a subset of the population and is the set of values you actually use in your estimation.

### Statistics:<br>
* It is  set of mathematical methods and tools that enable us to answer the important questions about data.<br>
* Two categories:<br>
  - <u>Descriptive:</u><br>
    * This offers method to summarized data by transforming raw observation into meaningful information that is easy to interpret and share.<br>
  - <u>Inferential:</u><br>
    * This offers the method to study the experiments done on small sample of chalk out of inferences to the entire popuation.



* <u>Descriptive Statistics:</u>  <br>

    - Measures of central tendency
    - Measures of Variability
    - Measure of shape
    - Percentiles
    - Frequency Distribution
    - Covariance and correlation 

* <u>Inferential Statistics:</u>

    - Probability Distribution
    - Central limit theorem
    - Regression Analysis
    - Hypothesis Testing
        * Z-test
        * T-test
        * Chi square test
