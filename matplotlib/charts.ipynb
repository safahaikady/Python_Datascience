### Introduction to matplotlib

```python
import matplotlib.pyplot as plt
import pandas as pd
```

#### Bar Plot

```python
course_enroll={
    "courses":['Java','Python','C++','JavaScript'],
    "enrolled":[25,15,18,10]
}
course_df=pd.DataFrame(course_enroll)
#create plot
plt.Figure(figsize=(9,5))
bars=plt.bar(course_df['courses'],course_df['enrolled'],color=["#883e6a",'#7e4887','#aa2382','#a42c44'],edgecolor='black',width=0.7)
#create title and labels
plt.title('Course Enrolled Students Overview',fontsize=15,pad=10,fontweight='bold')
plt.xlabel('Courses',fontsize=12)
plt.ylabel('Number of Students',fontsize=12)
plt.bar_label(bars,padding=1,fontweight='bold')
#Improve
plt.ylim(0,30)
plt.grid(axis='y',linestyle=':',alpha=0.3) #:,--
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```

#### Line Plot

```python
weekly_sales={
    "days":['Mon','Tues','Wed','Thu','Fri','Sat','Sun'],
    'first_week':[30000,30000,30000,44000,30000,30000,30000],
    'second_week':[30000,40000,30000,38000,30000,40000,30000],
}
sales=pd.DataFrame(weekly_sales)
plt.figure(figsize=(9,5))
#first week sales
plt.plot(sales["days"],sales['first_week'],color='red',ls="--",marker='*',label='First Week')#marker=*,o,d
#second week sales
plt.plot(sales["days"],sales['second_week'],color='blue',ls="--",marker='*',label='Second Week')
#Create title label
plt.title('Weekly Sales Overview',fontsize=15,fontweight='bold')
plt.xlabel('Days of week',fontsize=12)
plt.ylabel('Sales amount per day',fontsize=12)
plt.legend()
plt.tight_layout()
plt.show()
```

#### Scatter Plot

```python
student_age={
    "age":[20,22,21,23,20,22,22,23,20,22,25,21,22,23,25,21,23,20,21,20,21,23,20,20,24],
    "marks":[490,465,325,450,490,250,486,250,352,492,263,482,450,460,453,420,402,452,450,492,450,440,462,456,450]
}
std_df=pd.DataFrame(student_age)
plt.figure(figsize=(9,5))
plt.scatter(std_df['marks'],std_df['age'],color='blue',marker='*',s=100,alpha=0.3,label="marks")
#title
plt.title("Based on age check marks",fontweight="bold",fontsize=15)
plt.xlabel("Marks",fontsize=12)
plt.ylabel("student age",fontsize=12)
plt.legend()
plt.show()
```

### Pie Chart

```python
weekly_sales={
    "Brands":['Oneplus','Samsung','Apple','Vivo','Oppo'],
    "sales":[23,15,18,10,12]
}
sales_df=pd.DataFrame(weekly_sales)
plt.pie(sales_df['sales'],labels=sales_df['Brands'],explode=[0.1,0,0,0,0],autopct='%.2f%%',shadow=True,pctdistance=0.7,startangle=30)
plt.title("Weekly Sales Overview",fontsize=15,fontweight='bold')
plt.legend(loc='right',bbox_to_anchor=(1.5,0.8))
plt.show()
```

### Histogram

```python
temp_celsius=[28,29,30,31,31,30,29,28,27,26,25,24,24,24]
plt.hist(temp_celsius,bins=5,color='purple',alpha=0.8,edgecolor='pink')
plt.title("Temperature Overview",fontsize=15,fontweight='bold',pad=5)
plt.xlabel("Temperature",fontsize=12)
plt.ylabel("Frequency",fontsize=12)
plt.grid(axis='y',linestyle='--',alpha=0.5)  
plt.show()
```

### Violin chart

```python
#out of 100
marks=[10,20,30,40,50,60,62,64,65,67,70,80,90,95]
plt.violinplot(marks,showmeans=True,showmedians=True,facecolor='black')
plt.ylim(0,110)
plt.show()
```

### Stack Plot

```python
# Weekly Sales Data (in units)
weekly_sales = {
    "Days": ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"],
    "Week1": [5, 10, 30, 20, 35, 60, 80],
    "Week2": [30, 60, 30, 70, 80, 50, 30],
    "Week3": [10, 40, 50, 23, 69, 82, 43]
}
sales=pd.DataFrame(weekly_sales)
plt.stackplot(sales["Days"],sales['Week1'],sales['Week2'],sales['Week3'],labels=['Week1','Week2','Week3'])
plt.legend()
plt.show()
```
