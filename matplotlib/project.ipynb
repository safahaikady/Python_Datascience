```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#pie chart for each category in expense
df=pd.read_excel('expense.xlsx')
exp=pd.DataFrame(df)
df_plot=exp.groupby('Category')['Amount'].sum()
plt.pie(
    df_plot.values,labels=df_plot.index,autopct='%.2f%%',explode=[0,0.1,0,0,0],shadow=True,pctdistance=0.7,startangle=30
)
# df_plot.plot.pie(autopct='%.2f%%',explode=[0.1,0,0,0,0],shadow=True,pctdistance=0.7,startangle=30)
plt.legend(loc='right',bbox_to_anchor=(1.5,0.8))
plt.title("Expense Overview",fontsize=15,fontweight='bold')
```
